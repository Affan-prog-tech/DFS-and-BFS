```python
def smart_street_light_agent(light_condition):
    
    print(f"Current light condition: {light_condition}")
    
    if light_condition.lower() == "dark":
        action = "Turn Light ON"
    elif light_condition.lower() == "bright":
        action = "Turn Light OFF"
    else:
        action = "Invalid condition. No action taken."
        
    print(f"Selected action: {action}")
    print("-" * 30)


print("  Test Scenario 1    ")
smart_street_light_agent("Dark")

print("  Test Scenario 2  ")
smart_street_light_agent("Bright")
```

```python
rooms = {
    "A": "Unknown",
    "B": "Unknown",
    "C": "Unknown"
}

print("SMART ROOM MONITORING ROBOT")
print("----------------------------")

initial_data = {}

for room in rooms:
    condition = input("Enter condition of Room " + room + " (Normal/Warning/Critical): ")
    initial_data[room] = condition

start_room = input("Enter starting room (A/B/C): ").upper()

current_room = start_room

while True:

    print("\nCurrent Room:", current_room)

    condition = initial_data[current_room]

    print("Percept:", condition)

    rooms[current_room] = condition

    print("Decision: Store the observed condition")

    print("Memory:", rooms)

    if rooms["A"] == "Normal" and rooms["B"] == "Normal" and rooms["C"] == "Normal":
        print("Action: All rooms are Normal. Robot stops.")
        break

    if "Unknown" not in rooms.values():
        print("Action: All rooms have been observed.")
        break

    if current_room == "A":
        next_room = "B"
    elif current_room == "B":
        next_room = "C"
    else:
        next_room = "A"

    while rooms[next_room] != "Unknown":
        if next_room == "A":
            next_room = "B"
        elif next_room == "B":
            next_room = "C"
        else:
            next_room = "A"

    print("Decision: Room", next_room, "has not been observed yet.")

    print("Action: Move from Room", current_room, "to Room", next_room)

    current_room = next_room

print("\nFINAL INFORMATION")
print("-----------------")

for room in rooms:
    print("Room", room, "=", rooms[room])
```

```python
def food_delivery_route_selection():

    print("Food Delivery Route Selection Agent")

    def calculate_utility(time, distance, cost):
        utility = 100 - (time * 0.5) - (distance * 2) - (cost * 0.05)
        return utility

    for scenario in range(1, 3):

        print("\nInput Scenario", scenario)

        routes = {}

        for route in ["Route A", "Route B", "Route C"]:

            print("\nEnter information for", route)

            time = float(input("Time (minutes): "))
            distance = float(input("Distance (km): "))
            cost = float(input("Cost (Rs.): "))

            routes[route] = {
                "time": time,
                "distance": distance,
                "cost": cost
            }

        best_route = None
        highest_utility = float("-inf")

        print("\nRoute Evaluation")

        for route_name, data in routes.items():

            utility = calculate_utility(
                data["time"],
                data["distance"],
                data["cost"]
            )

            print("Route:", route_name)
            print("Time:", data["time"])
            print("Distance:", data["distance"])
            print("Cost:", data["cost"])
            print("Utility:", round(utility, 2))
            print()

            if utility > highest_utility:
                highest_utility = utility
                best_route = route_name

        print("Decision: The agent selects", best_route)
        print("Highest Utility:", round(highest_utility, 2))


if __name__ == "__main__":
    food_delivery_route_selection()
```

```python
def washing_machine_agent():

    print("Automatic Washing Machine")

    for scenario in range(1, 3):

        print("\nInput Scenario", scenario)

        clothes = input("Enter clothes condition (Dirty/Clean): ").strip().capitalize()
        water = input("Enter water level (Low/Medium/High): ").strip().capitalize()

        print("\nAgent Processing")
        print("Percept - Clothes Condition:", clothes)
        print("Percept - Water Level:", water)

        if clothes == "Clean":
            action = "Stop"
            decision = "Clothes are clean."

        elif clothes == "Dirty" and water == "Low":
            action = "Add Water"
            decision = "Water level is low."

        elif clothes == "Dirty" and water == "Medium":
            action = "Start Washing"
            decision = "Sufficient water is available."

        elif clothes == "Dirty" and water == "High":
            action = "Drain Water"
            decision = "Water level is too high."

        else:
            action = "Stop"
            decision = "Invalid input."

        print("Decision:", decision)
        print("Action:", action)


if __name__ == "__main__":
    washing_machine_agent()
```

```python
def travel_assistant():

    print("Travel Assistant")

    def calculate_utility(time, cost, comfort):
        utility = 100 - (time * 5) - (cost * 0.005) + (comfort * 3)
        return utility

    for scenario in range(1, 3):

        print("\nInput Scenario", scenario)

        transportation = {}

        for option in ["Bus", "Train", "Car"]:

            print("\nEnter information for", option)

            time = float(input("Travel Time (hours): "))
            cost = float(input("Cost (Rs.): "))
            comfort = float(input("Comfort (1-10): "))

            transportation[option] = {
                "time": time,
                "cost": cost,
                "comfort": comfort
            }

        best_option = None
        highest_utility = float("-inf")

        print("\nTransportation Evaluation")

        for option, data in transportation.items():

            utility = calculate_utility(
                data["time"],
                data["cost"],
                data["comfort"]
            )

            print("Option:", option)
            print("Time:", data["time"])
            print("Cost:", data["cost"])
            print("Comfort:", data["comfort"])
            print("Utility:", round(utility, 2))
            print()

            if utility > highest_utility:
                highest_utility = utility
                best_option = option

        print("Decision:", best_option, "is selected.")
        print("Highest Utility:", round(highest_utility, 2))


if __name__ == "__main__":
    travel_assistant()
```
