```python
rooms = {
    "A": " Dirty ",
    "B": " Dirty ",
    "C": " Dirty "
}

current_location = "A"


def vacuum_agent(location, status):

    if status == " Dirty ":
        return " Suck "

    else:
        if location == "A":
            return " Move Right "
        elif location == "B":
            return " Move Right "
        else:
            return " Move Left "


while " Dirty " in rooms.values():
    print("Current Location :", current_location)
    print("Room Status :", rooms[current_location])

    action = vacuum_agent(current_location, rooms[current_location])
    print("Agent Action :", action)
    print()

    if action == " Suck ":
        rooms[current_location] = " Clean "

    elif action == " Move Right ":
        if current_location == "A":
            current_location = "B"
        elif current_location == "B":
            current_location = "C"

    elif action == " Move Left ":
        if current_location == "C":
            current_location = "B"
        elif current_location == "B":
            current_location = "A"

print("All rooms are clean!")
```

```python
routes = {" Route A": {" time ": 20 ," cost ": 500} ," Route B": {" time ": 15 ," cost ": 700} ," Route C": {" time ": 30 ," cost ": 300}}
def calculate_utility ( time , cost ) :
    utility = 1000 - ( time * 20) - cost
    return utility
utilities = {}
for route , information in routes . items () :
    utility = calculate_utility (information [" time "] ,information [" cost "])
utilities [ route ] = utility
best_route = max(utilities ,key = utilities . get)
print (" Route Utilities :")
for route , utility in utilities . items () :
    print ( route , "=", utility )
    print (" Selected Route :", best_route )
    print (" Selected Utility :", utilities [ best_route ])
```
