```python
bill = float(input("Enter total bill: "))
people = int(input("Enter number of people: "))
tip_percentage = float(input("Enter tip percentage: "))

tip = bill * tip_percentage / 100
total_bill = bill + tip
per_person = total_bill / people

print("\n--- Bill Details ---")
print("Tip:", tip)
print("Total Bill:", total_bill)
print("Amount per person:", per_person)
```

```python
marks1 = float(input("Enter marks for Subject 1: "))
marks2 = float(input("Enter marks for Subject 2: "))
marks3 = float(input("Enter marks for Subject 3: "))

total = marks1 + marks2 + marks3
average = total / 3
percentage = (total / 300) * 100

print("\n--- Result ---")
print("Total Marks:", total)
print("Average Marks:", average)
print("Percentage:", percentage, "%")
```

```python
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
operator = input("Enter operator: ")

if operator == "+":
    result = num1 + num2
elif operator == "-":
    result = num1 - num2
elif operator == "*":
    result = num1 * num2
elif operator == "/":
    result = num1 / num2
else:
    result = "Invalid operator"

print("Result:", result)
```

```python
password = input("Enter your password: ")

has_number = False
has_uppercase = False

for character in password:
    if character.isdigit():
        has_number = True

    if character.isupper():
        has_uppercase = True

if len(password) >= 8 and has_number and has_uppercase:
    print("Password is strong")
else:
    print("Password is weak")

    if len(password) < 8:
        print("Password must be at least 8 characters long")

    if not has_number:
        print("Password must contain a number")

    if not has_uppercase:
        print("Password must contain an uppercase letter")
```

```python
import random

user = input("Enter rock, paper, or scissors: ")

computer = random.choice(["rock", "paper", "scissors"])

print("Computer chose:", computer)

if user == computer:
    print("It's a tie")

elif user == "rock" and computer == "scissors":
    print("You win")

elif user == "paper" and computer == "rock":
    print("You win")

elif user == "scissors" and computer == "paper":
    print("You win")

else:
    print("Computer wins")
```

```python
print("----- Food Menu -----")
print("1. Burger  Rs. 500")
print("2. Pizza   Rs. 800")
print("3. Fries   Rs. 300")
print("4. Biryani Rs. 400")

choice = int(input("Select an item (1-4): "))

if choice == 1:
    item = "Burger"
    price = 500
elif choice == 2:
    item = "Pizza"
    price = 800
elif choice == 3:
    item = "Fries"
    price = 300
elif choice == 4:
    item = "Biryani"
    price = 400
else:
    item = "Invalid"
    price = 0

if item == "Invalid":
    print("Invalid selection")
else:
    print("Selected item:", item)
    print("Price: Rs.", price)
```
